# **Cleaning**

```python
df = data.copy()
df.head()
```

```python
# drop unnecessary columns
df.drop(['job_id', 'salary_currency', 'employee_residence', 'job_description_length', 'benefits_score'], axis=1, inplace=True)
```

```python
df.head()
```

```python
# rename columns 
df.rename(columns={'salary_usd':'annual_salary'}, inplace=True)
```

```python
df.head()
```

## Feature Engineering


```python
df['monthly_salary'] = df['annual_salary'] / 12
df.head()
```

```python
df.experience_level = df.experience_level.apply(lambda x: 'Entry' if x == 'EN' else 'Mid' if x == 'MI' else 'Senior' if x == 'SE' else 'Executive')
df.experience_level
```

```python
df.employment_type = df.employment_type.apply(lambda x: 'Full Time' if x == 'FT' else 'Part Time' if x == 'PT' else 'Contract' if x == 'CT' else 'Freelance')
df.employment_type
```

```python
df.company_size = df.company_size.apply(lambda x: 'Small <50' if x == 'S' else 'Medium 50-250' if x == 'M' else 'Large >250')
df.company_size
```

```python
df['remote_state'] = df['remote_ratio'].apply(lambda x: 'No remote' if x == 0 else 'Hybrid' if x == 50 else 'Fully remote')
df.head()
```

```python
# extract each skill from the required_skills column
skill_set = set()
for item in df.required_skills.apply(lambda x: x.split(', ')):
    for skill in item:
        skill_set.add(skill)
```

```python
skill_set
```

```python
# create a new column for each skill
for skill in skill_set:
    df[skill] = df.required_skills.apply(lambda x: 1 if skill in x else 0)
```

```python
# extracting weekend & weekday
df.posting_date = df.posting_date.astype('datetime64[ms]')
```

```python
df['is_weekend'] = df.posting_date.dt.day_name().apply(lambda x: 'Weekend' if x in ['Friday', 'Saturday'] else 'Weekday')
```

```python
# month column
df['posting_month'] = df.posting_date.dt.month
df.posting_month
```

```python
df.head()
```

```python
# drop unnecessary columns
df.drop(['remote_ratio', 'required_skills'], axis=1, inplace=True)
```

## Saving Clean Data

```python
df.to_csv("ai_job_cleaned.csv", index=False)
```

## **loc & iloc**
- loc: Search by Real Index
- iloc: Search by Index
